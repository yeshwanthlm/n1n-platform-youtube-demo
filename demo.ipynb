```python
import os, time
from openai import OpenAI

client = OpenAI(
    api_key="sk-**************************",
    base_url="https://api.n1n.ai/v1",  # the only other change
)

MODELS = ["claude-opus-5-5", "gpt-6-astra", "gemini-3.8-flash"]
PROMPT = "Explain what an AWS API gateway is in two sentences, for a beginner with a simple example."

for model in MODELS:
    start = time.time()
    reply = client.chat.completions.create(
        model=model,
        messages=[{"role": "user", "content": PROMPT}],
        max_tokens=500,  # plenty for two sentences
    )
    secs = time.time() - start
    print(f"\n=== {model} | {secs:.1f}s | {reply.usage.total_tokens} tokens ===")
    print(reply.choices[0].message.content)
```
